```python
CAT, SCH = "capstone_five", "project"
PATIENTS_PATH = "/Volumes/capstone_five/project/landing/patients/patients/"

# 1. Read all patient CSV files (patients_10, 11, 12 together)
df = (spark.read
      .option("header", True)
      .option("inferSchema", True)
      .csv(PATIENTS_PATH))

# 2. Remove duplicate patients, in case the same patient appears in two files
df = df.dropDuplicates(["Id"])
print("Patients:", df.count())

# 3. Save as a Delta table
(df.write.mode("overwrite").option("overwriteSchema", "true")
   .saveAsTable(f"{CAT}.{SCH}.dim_patients"))

# 4. Check
display(spark.table(f"{CAT}.{SCH}.dim_patients")
        .select("Id", "FIRST", "LAST", "GENDER", "BIRTHDATE").limit(5))
```
